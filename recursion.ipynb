```python
def func(x,n):
    if n==0:
        return
    print(x)
    func(x,n-1)

func(15,4)
```

```python
# print 1 to n(head recursion : phle kaam hoga uske baad function call hoga)
def func(i,n):
    if i>n:
        return
    print(i)
    func(i+1,n)
func(1,4)
```

```python
#tail recusrion 
def func(i,n):
    if i>n:
        return
    func(i+1,n)
    print(i)

func(1,4)
```

```python
#n-1 using head 
def func(n):
    if n==0:
        return
    print(n)
    func(n-1)
func(4)
```

```python
#1-n using tail
def func(n):
    if n==0:
        return
    func(n-1)
    print(n)
func(4)
```

```python
#sum of n naturals using recursion parameterized
def func(sum,i,n):
    if i>n:
        print(sum)
        return
    func(sum+i,i+1,n)

func(0,1,5)
```

```python
#sum of n numbers using functional recursion 
def func(n):
    if n==1:
        return 1
    return n+func(n-1)
func(10)
```

```python
#FACTORIAL 
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n*fact(n-1)

fact(5)
```

```python
#reverse an array using recursion 
arr = [5,9,8,3,6,7,1,4,2]
def func(arr,l,r):
    if l>=r:
        return
    arr[l],arr[r]=arr[r],arr[l]
    func(arr,l+1,r-1)

func(arr,0,5)
print(arr)

```

```python
#string palindrome
s ="nitin"
def func(s):
    l=0
    r=len(s)-1
    while(l<r):
        if(s[l]!=s[r]):
            return False
        l+=1
        r-=1
    return True

print(func(s))

```
