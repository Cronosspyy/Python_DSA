```python
n = [1,2,3,4,5,5,2,2,1,3,4,3,1]
freq = {}
for i in range(0,len(n)):
    if n[i] in freq:
        freq[n[i]] += 1
    else:
        freq[n[i]] = 1

print(freq)
```

```python
nums = [1,2,3,4,5,5,2,2,1,3,4,3,1]
hashmap = {}
n = len(nums)
for i in range(0,n):
    hashmap[nums[i]] = hashmap.get(nums[i],0)+1
print(hashmap)
```
