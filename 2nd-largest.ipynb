```python
nums = [1,2,33,45,-3,23,57,88,98,54]
largest = float("-inf")
s_largest = float('-inf')
for i in range(len(nums)):
    if nums[i]>largest:
        largest = nums[i]
for j in range(len(nums)):
    if nums[j]>s_largest and nums[j]!=largest:
        s_largest = nums[j]

print(s_largest)
```

```python
#optimal
nums = [1,2,33,45,-3,23,57,88,98,54]
largest = float("-inf")
s_largest = float('-inf')
for i in range(0,len(nums)):
    if nums[i]> largest:
        s_largest = largest
        largest = nums[i]
    elif nums[i]> s_largest and nums[i]!= largest:
        s_largest = nums[i]

print(s_largest)
```
