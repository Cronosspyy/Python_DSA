```python
#brute force
n = [5,3,2,2,1,5,5,7,5,10]
m = [10,111,1,9,5,67,2]

for nums in m:
    count = 0 
    for x in n:
        if x==nums:
            count+=1
    print(count)
```

```python
#optimal
n = [5,3,2,2,1,5,5,7,5,10]
m = [10,111,1,9,5,67,2]
hashlist = [0]*11

for num in n:
    hashlist[num] += 1

for x in m:
    if x<1 or x>10:
        print(0)
    else:
        print(hashlist[x])
```

```python
#character hashing 
s = "azyxyyzaaaa"
q=['d','a','y','x']
hashlist = [0]*26

for ch in s:
    ascii_value = ord(ch)
    index = ascii_value - 97 
    hashlist[index] += 1

for x in q:
    ascii_valuee = ord(x)
    index = ascii_valuee - 97
    print(hashlist[index])
```
